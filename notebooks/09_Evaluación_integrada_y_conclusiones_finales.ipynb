### 9. Evaluación integrada y conclusiones finales

#### 9.1 Objetivo

El objetivo de este notebook es integrar los principales resultados obtenidos durante el proceso experimental y evaluar de forma conjunta la utilidad de los datos sintéticos para la estimación de la Probabilidad de Default.

La evaluación se estructura en tres dimensiones complementarias:

1. similitud estadística entre los datos reales y sintéticos;
2. utilidad predictiva de los datos sintéticos para el entrenamiento de modelos de Machine Learning;
3. consistencia explicativa entre los modelos entrenados con datos reales y sintéticos mediante SHAP.

La integración de estas dimensiones permite determinar en qué medida el conjunto sintético seleccionado, Gaussian Copula V2.1, preserva propiedades relevantes del conjunto original y puede utilizarse como alternativa para el desarrollo experimental de modelos de riesgo crediticio.

Esta etapa no incorpora nuevos procesos de generación o modelado, sino que consolida los resultados obtenidos en las etapas anteriores para responder al objetivo general y a los objetivos específicos del trabajo.

#### 9.2 Dimensión 1 — Similitud estadística

La primera dimensión evalúa hasta qué punto los datos sintéticos preservan las propiedades estadísticas observadas en el conjunto real.

La comparación entre Gaussian Copula V2.1, CTGAN V1-C y TVAE V1-C mostró que ningún sintetizador domina todas las propiedades evaluadas. Sin embargo, Gaussian Copula V2.1 presentó el mejor comportamiento agregado en términos de similitud marginal, mientras que CTGAN y TVAE conservaron mejor determinadas relaciones estructurales.

Los resultados muestran que Gaussian Copula V2.1 reproduce con mayor fidelidad las distribuciones marginales de las variables numéricas y categóricas, mientras que los modelos neuronales presentan ventajas en determinadas relaciones estructurales.

Este resultado evidencia que la similitud marginal y la preservación de dependencias multivariadas constituyen propiedades diferentes. Un sintetizador puede reproducir adecuadamente las distribuciones individuales de las variables sin conservar con la misma fidelidad todas las relaciones existentes entre ellas.

Por tanto, la similitud estadística no puede evaluarse mediante una única métrica y debe considerarse de forma multidimensional.

#### 9.3 Dimensión 2 — Utilidad predictiva

La segunda dimensión evalúa si los datos sintéticos permiten entrenar modelos capaces de mantener capacidad predictiva cuando son evaluados sobre observaciones reales.

| Modelo final       |      ROC-AUC |       PR-AUC |        Brier |
| ------------------ | -----------: | -----------: | -----------: |
| LightGBM Real      | **0.764858** | **0.253704** | **0.067340** |
| LightGBM Synthetic |     0.728760 |     0.206189 |     0.070604 |


El modelo LightGBM entrenado con Gaussian Copula V2.1 alcanzó aproximadamente el 95.28 % del valor de ROC-AUC y el 81.27 % del valor de PR-AUC obtenido por el modelo entrenado con datos reales.

Estos resultados muestran que el conjunto sintético conserva una parte importante de la capacidad predictiva necesaria para discriminar observaciones de mayor y menor riesgo.

No obstante, la comparación entre algoritmos mostró que esta utilidad no es independiente del modelo utilizado. Logistic Regression presentó la menor pérdida entre los escenarios Real y Synthetic, mientras que Random Forest mostró la mayor reducción de rendimiento.

Por tanto, la utilidad predictiva de los datos sintéticos debe interpretarse conjuntamente con el algoritmo downstream empleado.

#### 9.5 Evaluación integrada

La evaluación conjunta de las tres dimensiones muestra que Gaussian Copula V2.1 proporciona una representación sintética útil, aunque no equivalente al conjunto real.

Desde la perspectiva estadística, el sintetizador presenta una elevada similitud marginal, especialmente en variables categóricas y en una parte importante de las variables numéricas. No obstante, determinadas relaciones multivariadas no se preservan con la misma fidelidad.

Desde la perspectiva predictiva, los modelos entrenados con datos sintéticos mantienen una capacidad discriminativa relevante sobre observaciones reales. El modelo final LightGBM Synthetic alcanzó un ROC-AUC de 0.728760 frente a 0.764858 del modelo Real.

Desde la perspectiva explicativa, ambos modelos comparten un núcleo importante de variables relevantes y muestran una elevada conservación de la dirección de los principales efectos, aunque existen diferencias en la magnitud y contribución individual de determinadas variables.

Estos resultados muestran que las tres dimensiones de evaluación no son equivalentes ni necesariamente evolucionan de forma conjunta. Una elevada similitud marginal no garantiza la preservación completa de dependencias, y una elevada utilidad predictiva no implica una reproducción exacta de la estructura explicativa.

En consecuencia, la utilidad de los datos sintéticos debe evaluarse mediante un enfoque multidimensional que considere simultáneamente similitud estadística, capacidad predictiva y consistencia explicativa.

El objetivo general del trabajo consistió en evaluar la utilidad de los datos sintéticos para la estimación de la Probabilidad de Default mediante modelos de Machine Learning, considerando su similitud respecto a los datos reales, su utilidad predictiva y la consistencia de las explicaciones obtenidas mediante Inteligencia Artificial Explicable.

Los resultados obtenidos permiten concluir que, bajo el dataset, configuraciones y protocolo experimental evaluados, los datos sintéticos pueden constituir una alternativa útil para el desarrollo y experimentación de modelos de riesgo crediticio, aunque no deben considerarse un sustituto equivalente de los datos reales.

#### 9.7 Limitaciones

Los resultados del presente trabajo deben interpretarse considerando determinadas limitaciones.

- En primer lugar, el análisis se realizó sobre un único conjunto de datos público, Home Credit Default Risk, por lo que los resultados no pueden generalizarse automáticamente a otros portafolios, instituciones o contextos crediticios.

- En segundo lugar, la comparación entre sintetizadores se realizó bajo configuraciones experimentales específicas. Gaussian Copula fue objeto de un proceso iterativo de optimización, mientras que CTGAN y TVAE fueron utilizados como alternativas neuronales bajo configuraciones controladas pero sin una optimización exhaustiva de hiperparámetros. Por tanto, los resultados no constituyen evidencia de superioridad universal de una arquitectura generativa.

- En tercer lugar, la selección de sintetizadores utilizó LightGBM como modelo diagnóstico común. Los experimentos posteriores mostraron que la utilidad de los datos sintéticos puede depender del algoritmo predictivo utilizado.

- Asimismo, no se realizó una evaluación cuantitativa de privacidad, por lo que los resultados no permiten afirmar que los datos sintéticos garanticen protección frente a riesgos de reidentificación o divulgación.

- Finalmente, el análisis XAI se realizó sobre una muestra reproducible de 5 000 observaciones del conjunto real de prueba. Aunque este tamaño permite realizar un análisis global amplio, una evaluación futura podría analizar la estabilidad de las métricas SHAP ante diferentes tamaños y muestras.

Los resultados muestran que los datos sintéticos pueden preservar una parte relevante de las propiedades necesarias para el desarrollo de modelos de riesgo crediticio, aunque la fidelidad obtenida depende de la dimensión analizada.

Gaussian Copula V2.1 presentó el mejor equilibrio para el objetivo del estudio, destacando especialmente en similitud marginal y en capacidad de transferencia predictiva hacia observaciones reales.

El modelado final mostró que LightGBM entrenado con datos sintéticos alcanzó un rendimiento inferior al modelo entrenado con datos reales, pero mantuvo una capacidad discriminativa relevante.

El análisis mediante SHAP mostró adicionalmente que ambos modelos comparten un núcleo común de variables explicativas y una elevada consistencia en la dirección de los principales efectos. En conjunto, los resultados evidencian que los datos sintéticos no deben entenderse como una reproducción exacta del conjunto real, sino como una representación aproximada cuya utilidad depende de las propiedades que logra preservar.
