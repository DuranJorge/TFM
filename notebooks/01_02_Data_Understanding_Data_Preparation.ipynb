```python
'''
Este resultado es excelente y, además, te permite justificar una decisión metodológica importante en el TFM. La transformación logarítmica ha tenido el efecto esperado.

Interpretación del gráfico
Distribución original (izquierda)

Se observa que la distribución de AMT_INCOME_TOTAL presenta una asimetría positiva muy pronunciada. La mayoría de los clientes se concentra en niveles de ingreso relativamente bajos, mientras que un reducido número de observaciones alcanza valores extremadamente elevados (hasta 117 millones), generando una cola larga hacia la derecha.

Como consecuencia:

La mayor parte de la información queda comprimida en las primeras barras.
Es difícil visualizar la distribución real de los ingresos.
Los valores extremos dominan la escala del gráfico.
Distribución log-transformada (derecha)

Después de aplicar:

df["AMT_INCOME_TOTAL_LOG"] = np.log1p(df["AMT_INCOME_TOTAL"])

ocurren varias cosas positivas:

La distribución se vuelve mucho más simétrica.
La mayoría de los datos puede observarse claramente.
Los valores extremos siguen existiendo, pero dejan de dominar la escala.
La variable adquiere una forma cercana a una distribución normal (aunque no perfectamente normal).

Este es precisamente el objetivo de una transformación logarítmica.

¿Por qué utilizar log1p y no log?

En la memoria puedes justificarlo así:

Se utilizó la transformación log(1+x) (log1p) debido a que es una práctica estándar en variables financieras con fuerte asimetría positiva. Esta transformación reduce la influencia de los valores extremos y mantiene la posibilidad de procesar valores iguales a cero, evitando problemas matemáticos asociados al logaritmo natural.

¿Significa que debemos transformar todas las variables financieras?

Aquí es donde entra una decisión metodológica importante.

Opción A (la más sencilla)

Aplicar log únicamente a:

AMT_INCOME_TOTAL

porque fue la variable más problemática.

Opción B (la que yo recomendaría para tu TFM)

Evaluar las cuatro variables monetarias:

AMT_INCOME_TOTAL
AMT_CREDIT
AMT_ANNUITY
AMT_GOODS_PRICE

y aplicar la transformación solo si realmente reduce la asimetría.

Por ejemplo:

variables_financieras = [
    "AMT_INCOME_TOTAL",
    "AMT_CREDIT",
    "AMT_ANNUITY",
    "AMT_GOODS_PRICE"
]

for var in variables_financieras:

    original = df[var].skew()

    transformada = np.log1p(df[var]).skew()

    print(f"{var}")
    print(f"Asimetría original     : {original:.2f}")
    print(f"Asimetría transformada : {transformada:.2f}")
    print("-"*40)

Así no aplicas una transformación "porque sí", sino porque demuestras con evidencia que mejora la distribución.

Algo importante para tu investigación

Aquí conectamos con uno de los objetivos del TFM.

Tú vas a generar datos sintéticos.

Entonces surge una pregunta metodológica:

¿El generador debe entrenarse con la variable original o con la variable log-transformada?

Yo recomendaría:

Limpiar los datos.
Tratar los valores faltantes.
Corregir inconsistencias (DAYS_EMPLOYED, etc.).
Aplicar las transformaciones necesarias (como log1p en variables muy sesgadas).
Entrenar el generador de datos sintéticos con el dataset ya preprocesado.

La razón es que el modelo generativo aprenderá distribuciones más estables y representativas. Después, cuando evalúes la calidad de la síntesis, podrás comparar las distribuciones transformadas entre el conjunto real y el sintético de manera consistente.

Una recomendación metodológica para el TFM

Dado que uno de tus objetivos es comparar la calidad de la data real y la sintética, yo no eliminaría la variable original AMT_INCOME_TOTAL. Mantendría ambas durante el análisis:

AMT_INCOME_TOTAL (original), para documentar cómo era la distribución inicial.
AMT_INCOME_TOTAL_LOG (transformada), para el modelado y la generación de datos sintéticos.

Esto te permitirá mostrar en la memoria que el preprocesamiento mejoró la forma de la distribución y justificar por qué el pipeline utiliza la versión transformada sin perder la referencia al dato original. Es una práctica común en proyectos de ciencia de datos y hace que tu metodología sea más transparente y fácil de defender.
'''
```

##### 2.1.5 Clasificación de variables por estrategia de tratamiento

```python
#=========================================================
# CLASIFICACIÓN SEGÚN ESTRATEGIA DE TRATAMIENTO
#=========================================================

def clasificar_variable(row):

    # Variables especiales
    if row["Variable"] == "DAYS_EMPLOYED":
        return "Valor especial"

    # Identificadores
    if row["Variable"].startswith("SK_ID"):
        return "Identificador"

    # Variables constantes
    if row["Constante"] == "Sí":
        return "Eliminar"

    # Valores faltantes
    if row["% Nulos"] == 0:
        return "Sin tratamiento"

    elif row["% Nulos"] <= 10:
        return "Imputación simple"

    elif row["% Nulos"] <= 50:
        return "Evaluar importancia"

    else:
        return "Conservar temporalmente"

quality_df["Estrategia"] = quality_df.apply(
    clasificar_variable,
    axis=1
)

quality_df.head()
```

#### 2.6 Análisis de variables categóricas y estrategia de codificación

Debido a que el conjunto preparado contiene variables categóricas, se analiza su cardinalidad con el propósito de identificar las estrategias de representación que podrían utilizarse posteriormente durante el modelado predictivo.

En esta etapa no se aplica ninguna transformación definitiva sobre las variables categóricas. El dataset conserva sus categorías originales, lo que permite su posterior utilización en los procesos de generación de datos sintéticos. La codificación requerida por los modelos de Machine Learning se realizará posteriormente dentro del pipeline de modelado, evitando modificar el conjunto preparado utilizado como entrada de los sintetizadores.

##### 2.6.1 Identificar variables categóricas

```python
categorical_cols = df_selected.select_dtypes(include=["object"]).columns.tolist()

print(f"Número de variables categóricas: {len(categorical_cols)}")

categorical_cols
```

##### 2.6.2 Analizar cardinalidad

```python
cardinality = pd.DataFrame({

    "Variable": categorical_cols,

    "Categorias": [

        df_selected[col].nunique()

        for col in categorical_cols

    ]

})

cardinality = cardinality.sort_values(

    "Categorias",

    ascending=False

)

cardinality
```

##### 2.6.3 Clasificación

```python
def tipo_encoding(x):

    if x <= 2:
        return "Label Encoding"

    elif x <= 10:
        return "One-Hot"

    else:
        return "Alta Cardinalidad"

cardinality["Encoding"] = cardinality["Categorias"].apply(tipo_encoding)

cardinality
```

```python
#Reemplaza la celda anterior
def clasificar_cardinalidad(x):
    if x <= 2:
        return "Binaria"
    elif x <= 10:
        return "Baja cardinalidad"
    else:
        return "Alta cardinalidad"

cardinality["Clasificacion"] = (
    cardinality["Categorias"]
    .apply(clasificar_cardinalidad)
)

cardinality
```

La clasificación anterior tiene carácter descriptivo y permite caracterizar la complejidad de las variables categóricas. La estrategia definitiva de codificación no se aplica en esta etapa, sino dentro del pipeline correspondiente a cada experimento de Machine Learning.

Para la generación de datos sintéticos se conservan las categorías originales, permitiendo que los sintetizadores modelen directamente la naturaleza categórica de estas variables mediante su metadata.
